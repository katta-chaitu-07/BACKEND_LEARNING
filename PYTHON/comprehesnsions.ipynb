## Python comprehensions

```python
# with out compression

# finding the squares of a number

number = [1,2,3,4,5]

squares = []

for num in number:

    squares.append(num*num)

print(squares)    
```

## useing the compression

```python
number = [1,2,3,4,5]

squares = [num*num  for num in number]

print(squares)
```

## Problem 1: Extract even numbers

```python
numbers = [10, 15, 20, 25, 30, 35, 40, 45, 50]

even = [num for num in numbers if num %2 == 0]

print(even)
```

## conditional comprehesnsion

```python
names = ["Chaitu", "Ram", "Krishna", "Sai", "Arjun"]

valid_names = [x for x in names if len(x) > 4]

print(valid_names)
```

Task: Create a new list containing numbers that satisfy both conditions:
1. Number must be greater than 20.
2. Number must be divisible by 5.



```python
numbers = [12, 7, 25, 40, 33, 18, 50, 9, 60]

number_list = [num for num in numbers if num > 20 and num % 5 == 0]

print(number_list)
```

## Distionary comprehension

```python
names = ["Chaitu", "Ram", "Krishna"]
marks = [85, 90, 95]

student = {}

for x in range(len(names)):

    student[names[x]] = marks[x]

print(student)    
  
```

```python
names = ["Chaitu", "Ram", "Krishna"]
marks = [85, 90, 95]

std = {names[i] : marks[i] for i in range(len(names))}

print(std)
```

## ZIP

```python
names = ["Chaitu", "Ram", "Krishna"]
marks = [85, 90, 95]

student = {name : mark
           for name , mark in zip(names,marks)
           }

print(student)
```

## Practical exercise

```python
products = ["Laptop", "Mobile", "Tablet", "Keyboard"]

prices = [65000, 30000, 25000, 2000]


# using the dictonary comprehension

print("using the dictonary comprehension \n")

student_dict = {

    products[i] : prices[i]

    for i in range(len(products))
}

print(student_dict)

print()

print("dictionary comprehension using the zip method \n ")


product_dict = {

    product : price 

    for product, price in zip(products,prices)
}


print(product_dict)
```

## Nested comprehensions

```python
for i in range(1, 4):
    for j in range(1, 4):
        print(i, j)
```

```python
results = [(i,j) for i in range(1,4) for j in range(1,4)]

print(results)
```

## practical example (Multiplication table)

```python
table = []

for i in range(1,5):

    row = []

    for j in range(1,5):

        row.append(i*j)

    table.append(row)

print(table)        
```

```python

table = [[i*j for i in range(1,6)] for j in range(1,6)]

print(table)
```

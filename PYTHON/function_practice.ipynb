## Positional Arguments

```python
def positional(name, age , percentage):

    print(f'Student_name : {name}')

    print(f'Student_age : {age}')

    print(f'Student_percentage : {percentage}')

positional("Chaitu",20 , 90)    
```

```python

# In Python we have 4 types of functioanl arguments 

# positonal argumenes
# Default arguments
# variable positonal arguments (*args)
# variable keyword argements (**kargs)


def calculate_student_result(name, age = 20, *marks, **additonal_information):

    print(f'student_name : {name}')

    print(f'Age : {age}')

    total = 0

    for x in marks:

        total += x

    print(f'Total_marks : {total}')

    print(f'Number of subjects : {len(marks)}')

    print(f"Average of marks : {total / len(marks)}")

    print(f"Additional information : {additonal_information}")  


calculate_student_result("Chaitu",
                         21,
                         90,
                         89,
                         75,
                         82,
                         80,
                         branch = "AI & ML",
                         university = "GITAM"
                         )
```

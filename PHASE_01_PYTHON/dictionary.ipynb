# Dictionary 

```python
student = {
    "id" : 62,
    "name" : "chaitanya",
    "branch" : "AIML"
}

print(student)
```

## Access dictionary values

```python
# we access the values using the keys 
student = {
    "id" : 62,
    "name" : "chaitanya",
    "branch" : "AIML"
}

print(student["id"])
print(student["name"])

#Rules 

# keys must be unique 
# value may be duplicate (no problem)

```

## get()

```python
# use get when we are not sure about the key is exist ,
#  we can handle this situation safely by using the get()


print(student.get("phone","not exist"))
```

## loop through a dictionary


```python
for key in student:
    print(key,student[key])
```

## Add a key to the dict

```python
# phone key is not exist in the student dict 

student["Phone"] = 123456789 # here phone(key) not exist , that why's it created automatically

for key in student:
    print(key,student[key])
```

# update the existing value in key

```python
print(student)

# i update the branch from AIML to CSE

student["branch"] = "CSE"

print("After updataion")

print(student)
```

## items()

```python
print(student.items())

for key,value in student.items():
    print(key ," : ", value)
```

# update()

```python
print(student)

student.update({"branch":"AIML"})

print(student)
```

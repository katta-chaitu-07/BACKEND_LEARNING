## Decorators 

```python
# Decorators used to add the extra behaviour to a function without changing the original code 

def great():

    print("Hello world")

great()    
```

## Concept A : ( A function is a value)

```python
def great():

    print("Hello chaitu , you are working on the decorators in python ! ")

x = great

x()

great()

# Note 

# great() is a function 

# x = great // assign that function to x
# 
# x = great() means we are calling the function ( function call) 
```

## concept B : passing a funciton as a argument

```python
def great():

    print("Hello chaitu , from the great function !")

def execute(fun):

    print("Hello chaitu, from the execute function !")

    fun()

    print("Hello chaitu, from the execute function !") 

execute(great)       


# great() is a function
# 
# execute(fun) this accepst the argument fun,
# 
# execute(great) , this is a function that passes the great as a argument  
```

## concept c : Returning a function 

```python

def outer():

    print("This is outer function ")

    def inner():

        print("This is inner function : ")

    return inner

    print("outer")

x = outer()

x()
```

## using the @ keyword

```python

def decarator(fun):

    def wrapper():

        print("Before ")
        fun()
        print("After")
    return wrapper

@decarator
def great():

    print("Hello chaitu")

great()        
```
